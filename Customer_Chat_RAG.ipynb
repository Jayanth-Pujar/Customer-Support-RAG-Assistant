```python
from langchain_community.document_loaders import PyPDFLoader

print("PyPDFLoader imported successfully")
```

```python
# Load PDF
loader = PyPDFLoader('/content/drive/MyDrive/ML_Projects/Customer_Chat_RAG/customer_support_policy.pdf')

documents = loader.load()

print("Number of pages:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50
)

chunks = text_splitter.split_documents(documents)
```

```python
for i, chunk in enumerate(chunks):
  print(f"\n--- Chunk {i+1} ---")
  print(chunk.page_content)
```

```python
!pip install -q sentence-transformers
```

```python
from langchain_community.embeddings import HuggingFaceBgeEmbeddings

embeddings = HuggingFaceBgeEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully")
```

```python
!pip install -q faiss-cpu
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

print("Vector database created successfully")
```

```python
question = "How many days do I have to return a product?"
```

```python
results = vectorstore.similarity_search(
    question,
    k=3
)

for i, result in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(result.page_content)
```

```python
import requests

def ask_question(question):

    # 1. Retrieve relevant chunks from FAISS
    results = vectorstore.similarity_search(
        question,
        k=3
    )

    # 2. Combine retrieved chunks into context
    context = "\n\n".join(
        [result.page_content for result in results]
    )

    # 3. Create the prompt
    prompt = f"""
You are a customer support assistant.

Answer the question using only the information provided in the context.

If the answer is not available in the context, say:
"I don't have enough information in the provided documents."

Context:
{context}

Question:
{question}

Answer:
"""

    # 4. Send prompt to Ollama
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "llama3.2:1b",
            "prompt": prompt,
            "stream": False
        }
    )

    # 5. Extract the answer
    answer = response.json()["response"]

    return answer
```

```python
print(ask_question("How many days do I have to return a product?"))
```
